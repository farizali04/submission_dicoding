## Menentukan Pertanyaan Bisnis

**SMART Question** adalah sebuah framework untuk merumuskan pertanyaan secara terstruktur agar memeperoleh informasi yang mendalam.

* **Spesific (Spesifik)**
  * Pertanyaan harus jelas, fokus pada sebuah topik tertentu, dan tidak bermakna ganda. Hindari pertanyaan yang terlalu luas.
    * Salah: Bagaimana cara meningkatkan penjualan?
    * Benar: Faktor apa saja yang memengaruhi penurunan penjualan produk kategori elektronik di wilayah Jakarta selama kuartal terakhir?
* **Measurable (Terukur)**
  * Pertanyaan harus bisa dijawab dengan angka atau matrix yang konkret, harus tahu apa yang akan dihitung.
    * Salah: Apakah pelanggan senang dengan layanan kita?
    * Benar: Berapa skor rata-rata Customer Satisfaction untuk layanan purna jual bulan ini dibandingkan bulan lalu?
* **Action-Oriented (Berorientasi Aksi)**
  * Hasil dari pertanyaan harus bisa memberikan arahan untuk melakukan tindakan nyata. Jika pertanyaan terjawab, stakeholder harus tahu apa langkah selanjutnya.
    * Salah: Mengapa orang suka berbelanja?
    * Benar: Fitur apa pada aplikasi yang paling sering digunakan sebelum pengguna memutuskan untuk melakukan checkout?
* **Relevant (Relevan)**
  * Hasil dari pertanyaan harus sejalan dengan tujuan utama bisnis atau masalah yang sedang dihadapi.
    * Salah: Menanyakan tentang stok gudang saat masalah utamanya adalah efektivitas kampanye media sosial.
    * Benar: Apakah kampanye iklan di Instagram memberikan Return on Ad Spend (ROAS) yang lebih tinggi dibandingkan iklan di TikTok?
* **Time-bound (Terikat Waktu)**
  * Pertanyaan harus ada batasan waktu yang jelas agar analisis memiliki konteks yang tepat.
    * Salah: Berapa banyak pengguna baru kita?
    * Benar: Berapa tingkat pertumbuhan pengguna baru secara bulanan (Month-over-Month) sepanjang tahun 2025?

**Contoh pertanyaan bisnis yang memenuhi seluruh elemen SMART**

- **Specific**: Fokus pada "penurunan conversion rate" untuk "aplikasi Android" di "Jabodetabek". Bukan sekadar penjualan turun.

- **Relevant**: Penurunan konversi saat Flash Sale adalah masalah kritis bagi bisnis retail/e-commerce.
- **Time-bound**: Dibatasi pada periode spesifik "Flash Sale Maret 2026".

## Visualization & Explanatory Analysis

### Pertanyaan 1:
Bagaimana pola jumlah penyewaan sepeda (cnt) berdasarkan jam (hr) pada hari kerja vs akhir pekan di musim panas tahun 2012, sehingga manajemen dapat mengatur penambahan staf dan sepeda pada jam puncak?

```python
# Filter data tahun 2012 pada musim panas
df_2012_summer = df_clean[(df_clean['yr'] == 1) & (df_clean['season'] == 3)].copy()

# Membuat mapping untuk workingday : 0 = Weekend/Holiday, 1 = Weekday
df_2012_summer['day_type'] = df_2012_summer['workingday'].map({0 : 'Weekend/Holiday', 1 : 'Weekday'})

# Agregasi rata-rata cnt per jam dan tipe hari
hourly_avg = df_2012_summer.groupby(['hr', 'day_type'])['cnt'].mean().reset_index()
display(hourly_avg.head(10))

# Visualiasi 1 : Line Plot untuk pola Jam sibuk
plt.figure(figsize=(14, 6))
sns.lineplot(data = hourly_avg, x='hr', y='cnt', hue='day_type', marker='o')
plt.title('Rata-rata Penyewaan Sepeda per Jam pada Musim Panas tahun 2012')
plt.xlabel('Jam')
plt.ylabel('Rata-rata Penyewaan Sepeda')
plt.xticks(range(0, 24))
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend(title='Tipe Hari')
plt.show()

# Identifikasi Jam Puncak pada Weekday
weekday_data = hourly_avg[hourly_avg['day_type'] == 'Weekday']
peak_weekday = weekday_data.loc[weekday_data['cnt'].idxmax()]

# Identifikasi Jam Puncak pada Weekend/Holiday
weekend_data = hourly_avg[hourly_avg['day_type'] == 'Weekend/Holiday']
peak_weekend = weekend_data.loc[weekend_data['cnt'].idxmax()]

# Print
print(f"\nJam puncak weekday: {peak_weekday['hr']}:00 dengan rata-rata {peak_weekday['cnt']:.0f} sepeda")
print(f"Jam puncak weekend: {peak_weekend['hr']}:00 dengan rata-rata {peak_weekend['cnt']:.0f} sepeda")
```

### Pertanyaan 2:
Apakah pengaruh cuaca buruk terhadap jumlah pengguna casual dan registered berbeda secara signifikan pada jam sibuk pagi (6-9) dan sore (17-19) di tahun 2012?

```python
# Pengaruh cuaca buruk pada casual dan registered di jam sibuk

# Jam sibuk untuk pagi hari : 6 - 9, Jam sibuk untuk sore hari : 17 - 19
df_2012 = df_clean[(df_clean['yr'] == 1)].copy()
df_2012['peak_hour'] = 'Lainnya'
df_2012.loc[df_2012['hr'].between(6, 9), 'peak_hour'] = 'Pagi (6-9)'
df_2012.loc[df_2012['hr'].between(17, 19), 'peak_hour'] = 'Sore (17-19)'
df_2012 = df_2012[df_2012['peak_hour'] != 'Lainnya']

# Mengkategorikan cuaca : baik (weathersit 1,2), buruk (weathersit 3,4)
df_2012['weather_good'] = df_2012['weathersit'].apply(lambda x: 1 if x <= 2 else 0)

# Menghitung rata-rata casual dan registered
peak_agg = df_2012.groupby(['peak_hour', 'weather_good'])[['casual', 'registered']].mean().reset_index()
print("=== Tabel Agregasi: Rata-rata Casual & Registered pada Jam Sibuk (Cuaca Baik/Buruk) ===")
display(peak_agg)

# Menampilkan Hasil dari penyewaan sepeda pada kondisi cuaca tertentu
fig, axes = plt.subplots(1, 2, figsize=(14,5))

df_2012['weather_label'] = df_2012['weather_good'].map({
    1: 'Baik',
    0: 'Buruk'
})

sns.barplot(data=df_2012, x='peak_hour', y='casual', hue='weather_label', ax=axes[0], palette='Set2')
axes[0].set_title('Casual: Cuaca Baik vs Buruk')

sns.barplot(data=df_2012, x='peak_hour', y='registered', hue='weather_label', ax=axes[1], palette='Set2')
axes[1].set_title('Registered: Cuaca Baik vs Buruk')
```

**Insight:** (Opsional)
- Insight dari EDA Pertanyaan 1 (Pola Jam Sibuk di Musim Panas 2012)
  - Hari kerja (weekday):
  Dua puncak penyewaan: jam 7–8 pagi (rata-rata 440 sepeda/jam) dan jam 17–18 sore (430 sepeda/jam).
  Ini adalah jam berangkat dan pulang kerja (pola komuter).
  - Akhir pekan / hari libur (weekend):
  Hanya satu puncak di jam 12–14 siang (350 sepeda/jam).
  Pola ini menunjukkan aktivitas rekreasi (bersepeda di siang hari).
  - Perbedaan volume:
  Jam puncak weekday lebih tinggi sekitar 90–100 sepeda/jam dibanding puncak weekend.
  Ini mengindikasikan bahwa registered (komuter) lebih mendominasi di hari kerja.
  - Jam sepi:
  Baik weekday maupun weekend, jam 0–5 dini hari sangat sepi (< 50 sepeda/jam).

- Insight dari EDA Pertanyaan 2 (Pengaruh Cuaca Buruk pada Casual vs Registered di Jam Sibuk)
  - Penurunan drastis pada pengguna casual:
  Pagi (6–9): casual turun dari 42 → 12 (turun 71%).
  Sore (17–19): casual turun dari 55 → 18 (turun 67%).
  - Penurunan lebih kecil pada pengguna registered:
  Pagi: registered turun dari 360 → 222 (turun 38%).
  Sore: registered turun dari 430 → 280 (turun 35%).
  - Kesimpulan utama:
  Casual jauh lebih sensitif terhadap cuaca buruk dibanding registered.
  Registered cenderung tetap bersepeda meskipun cuaca tidak mendukung (karena kebutuhan komuter).
  - Implikasi:
  Cuaca buruk di jam sibuk lebih berdampak pada kehilangan pendapatan dari segmen casual.

## Analisis Lanjutan (Opsional)

```python
# Analisis Lanjutan - Pola per jam untuk tiap kondisi cuaca
print("=== ANALISIS LANJUTAN: Pola per jam berdasarkan kondisi cuaca (2012) ===")

df_2012 = df_clean[df_clean['yr'] == 1].copy()
df_2012['weather_label'] = df_2012['weathersit'].map({1: 'Cerah', 2: 'Berawan', 3: 'Hujan Ringan', 4: 'Hujan Lebat'})

hourly_weather = df_2012.groupby(['hr', 'weather_label'])['cnt'].mean().reset_index()

plt.figure(figsize=(14,6))
sns.lineplot(data=hourly_weather, x='hr', y='cnt', hue='weather_label', marker='o')
plt.title('Pola Penyewaan per Jam Berdasarkan Cuaca (2012)')
plt.xlabel('Jam')
plt.ylabel('Rata-rata cnt')
plt.xticks(range(0,24))
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend(title='Cuaca')
plt.tight_layout()
plt.show()
```
